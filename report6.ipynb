### Выполнили Климовский Дмитрий и Грудинин Михаил 23Б16

```python
import numpy as np
```

```python
def jacobi_method(A, tol=1e-10, max_iterations=100):
    n = A.shape[0]
    for iteration in range(max_iterations):
        # Найдем максимальный внедиагональный элемент
        off_diagonal_max = 0
        p, q = 0, 1
        for i in range(n):
            for j in range(i+1, n):
                if abs(A[i, j]) > off_diagonal_max:
                    off_diagonal_max = abs(A[i, j])
                    p, q = i, j
        # Проверка на достижение точности
        if off_diagonal_max < tol:
            break
        # Вычисление угла вращения
        if A[p, p] != A[q, q]:
            tau = (A[q, q] - A[p, p]) / (2 * A[p, q])
            t = np.sign(tau) / (abs(tau) + np.sqrt(1 + tau**2))
        else:
            t = 1
        c = 1 / np.sqrt(1 + t**2)
        s = t * c
        # Обновление матрицы A
        for i in range(n):
            if i != p and i != q:
                A_ip = A[i, p]
                A_iq = A[i, q]
                A[i, p] = c * A_ip - s * A_iq
                A[p, i] = A[i, p]
                A[i, q] = c * A_iq + s * A_ip
                A[q, i] = A[i, q]
        A_pp = A[p, p]
        A_qq = A[q, q]
        A_pq = A[p, q]
        A[p, p] = c**2 * A_pp + s**2 * A_qq - 2 * s * c * A_pq
        A[q, q] = s**2 * A_pp + c**2 * A_qq + 2 * s * c * A_pq
        A[p, q] = 0
        A[q, p] = 0
    
    eigenvalues = np.diag(A)
    return eigenvalues
```

Наша матрица:

```python
A = np.array([[4, 1, 2],
              [1, 6, 1],
              [2, 1, 9]], dtype=np.float64)
```

Получим решение методом Якоби и готовым решением от библиотеки numpy

```python
eigenvalues = jacobi_method(A)
print(f"Собственные значения матрицы, полученные методом Якоби: {eigenvalues}")

eigenvalues_true, _ = np.linalg.eig(A)
print(f"Истинные собственные значения матрицы: {eigenvalues_true}")
```
